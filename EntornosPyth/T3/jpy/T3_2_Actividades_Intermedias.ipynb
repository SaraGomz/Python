## Actividad 7. Proyecto inicial

Crea un proyecto denominado **`agenda_personal`** que incluya:

```
agenda_personal/
├── README.md
├── .gitignore
└── main.py
```

El archivo `main.py` deberá mostrar un mensaje personalizado.

### 7.1 Crear la carpeta y los archivos

````python
from pathlib import Path

proyecto = Path("agenda_personal")
proyecto.mkdir(exist_ok=True)

for nombre in ["README.md", ".gitignore", "main.py"]:
    (proyecto / nombre).touch()

print("Contenido de la carpeta:")
for elemento in sorted(proyecto.iterdir()):
    print(" -", elemento.name)
````

### 7.2 Escribir `main.py` (personaliza el mensaje)

````python
%%writefile agenda_personal/main.py
nombre = "TU NOMBRE"
print(f"Hola, {nombre}. Bienvenido/a a tu agenda personal.")
````

### 7.3 Escribir `README.md` (complétalo)

````python
%%writefile agenda_personal/README.md
# Agenda personal

Describe aquí el objetivo del proyecto.

## Cómo ejecutarlo

    python main.py
````

### 7.4 Escribir `.gitignore`

````python
%%writefile agenda_personal/.gitignore
__pycache__/
.vscode/
.venv/
````

### 7.5 Ejecutar el proyecto

````python
!cd agenda_personal && python main.py
````
