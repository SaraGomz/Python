## Actividad 12. Reflexión profesional

Redacta aproximadamente **una página** respondiendo a la siguiente pregunta:

> **¿Por qué los desarrolladores profesionales utilizan herramientas como Git, VS Code y pytest incluso para proyectos relativamente pequeños?**

Argumenta tu respuesta utilizando ejemplos.

✍️ **Tu reflexión:**

Los desarrolladores profesionales usan herramientas como Git, VS Code y pytest incluso en proyectos pequeños porque optimizan la productividad, garantizan la calidad del código y previenen errores.

En primer lugar, Git actúa como una red de seguridad e historial del código. En lugar de crear copias manuales como script_v1.py o script_final.py, un comando como git commit -m guarda un punto de control exacto, permitiendo experimentar en ramas aisladas y revertir a un estado funcional si algo sale mal. Por su parte, VS Code centraliza el trabajo al integrar el editor, la terminal y herramientas de depuración en una sola interfaz limpia por ejemplo, permite corregir un error detectado en la terminal integrada sin necesidad de cambiar de ventana. Finalmente, pytest automatiza la verificación del programa. En vez de comprobar manualmente los resultados con llamadas a print(), un simple assert sumar(2, 2) == 4 ejecutado por pytest valida al instante que los cambios recientes no hayan generado regresiones.

En conclusión, la combinación de Git, VS Code y pytest ahorra tiempo y garantiza la organización y calma de cualquier proyecto, sin importar su tamaño.

### P.1 Crear la estructura

````python
from pathlib import Path

raiz = Path("mi_primer_proyecto")
(raiz / "src").mkdir(parents=True, exist_ok=True)
(raiz / "tests").mkdir(exist_ok=True)

for ruta in ["README.md", ".gitignore", "pytest.ini", "src/main.py", "tests/test_main.py"]:
    (raiz / ruta).touch()

for elemento in sorted(raiz.rglob("*")):
    print(elemento.relative_to(raiz.parent))
````

### P.2 Programa inicial: `src/main.py`

Muestra nombre del alumno, fecha actual y mensaje de entorno operativo. **Escribe tu nombre.**

````python
%%writefile mi_primer_proyecto/src/main.py
"""Programa inicial del proyecto integrador."""
from datetime import date

NOMBRE_ALUMNO = "TU NOMBRE"


def construir_mensaje(nombre, fecha):
    """Devuelve el mensaje de bienvenida con nombre y fecha."""
    return (
        f"Alumno: {nombre}\n"
        f"Fecha: {fecha.strftime('%d/%m/%Y')}\n"
        "El entorno está operativo."
    )


def main():
    print(construir_mensaje(NOMBRE_ALUMNO, date.today()))


if __name__ == "__main__":
    main()
````

### P.3 Prueba automatizada: `tests/test_main.py`

````python
%%writefile mi_primer_proyecto/tests/test_main.py
from datetime import date

from main import construir_mensaje


def test_mensaje_contiene_nombre_fecha_y_estado():
    mensaje = construir_mensaje("Ana", date(2026, 1, 15))
    assert "Ana" in mensaje
    assert "15/01/2026" in mensaje
    assert "operativo" in mensaje
````

Para que pytest encuentre el código de `src/`, añadimos un pequeño archivo de configuración:

````python
%%writefile mi_primer_proyecto/pytest.ini
[pytest]
pythonpath = src
````
