## Actividad 2. Entorno global vs entorno virtual

Explica las diferencias, ventajas e inconvenientes de cada opción:
- Opción A: Instalar todas las bibliotecas en el sistema global.
- Opción B: Utilizar entornos virtuales independientes.

### Respuesta

Opción A
Ventajas: es rápido y directo para scripts sencillos o cuando estas empezando, a parte las bibliotecas son accesibles desde cualquier ubicación
Desventajas: Pueden haber problemas de versiones entre proyectos, en algunos sistemas operativos usan python y sus propias bibliotecas internas modificar paquetes a nivel global puede romper estas herramientas y por último, es dificil de compartir a otros desarrolladores ya que tienen que tener las exactas mismas dependencias instaladas

Opción B
Ventajas: cada proyecto está aislado y tiene las versiones que necesita sin molestar a otros proyectos, es fácil la creación de archivos de configuración que nos permiten duplicar el entorno en otra máquina y si algo se corrompe se elimina la carpeta del entorno y creas otra nueva sin afectar a nada más
Desventajas: tienes que saber comandos adicionales para crear, activar y desactivar los entornos y tienes que acordarte de activar el entorno correcto cada vez que vayas a trabajar en un proyecto específico desde la terminal
