## Actividad 11. Comparativa de herramientas modernas

Elabora una tabla comparativa entre:
- venv
- pip
- Poetry
- uv
Para cada herramienta indica:
- Objetivo.
- Ventajas.
- Limitaciones.
- Contexto de uso.

Finaliza con una recomendación razonada.

| Herramienta | Objetivo | Ventajas | Limitaciones | Contexto de uso (VS Code / Jupyter) |
| :--- | :--- | :--- | :--- | :--- |
| **`venv`** | Crear entornos virtuales aislados nativos en Python | Viene preinstalado con Python es ligero y directo. | No gestiona paquetes por sí mismo ni genera archivos de bloqueo avanzados de forma nativa | Ideal para scripts rápidos, aprendizaje o cuando quieres configurar manualmente un kernel de Jupyter sin dependencias complejas |
| **`pip`** | Instalar, actualizar y desinstalar paquetes desde PyPI | Es universal y tiene una sintaxis muy simple e intuitiva | No maneja entornos virtuales por sí solo ni resuelve conflictos eficientemente en proyectos grandes | Se usa siempre dentro de un venv para instalar paquetes puntuales |
| **`Poetry`** | Gestión integral del ciclo de vida del proyecto | Excelente resolución de dependencias | A veces la integración estricta de su entorno virtual choca | Proyectos medianos y grandes en equipo o librerías que se van a publicar requiere registrar manualmente el entorno virtual |
| **`uv`** | Gestor universal de Python escrito en Rust | Velocidad extrema | Al ser una herramienta más moderna, los equipos tradicionales aún están en proceso de adopción completa | Excelente para ciencia de datos crea entornos y sincroniza paquetes al instante, integrándose de forma muy limpia |


La velocidad de descarga y resolución de dependencias de uv elimina los tiempos de espera muertos, por otra parte uv crea automáticamente la carpeta .venv en la raíz del proyecto. Al abrir un archivo .ipynb en VS Code, este detecta de inmediato el entorno virtual generado por uv para que puedas seleccionarlo
